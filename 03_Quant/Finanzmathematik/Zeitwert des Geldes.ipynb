# Future Value und Present Value von Geld

## Der Zeitwert des Geldes mit Python

Der Zeitwert des Geldes beschreibt das grundlegende Prinzip, dass ein Geldbetrag heute einen anderen Wert hat als derselbe Geldbetrag in der Zukunft. Der Grund dafür ist, dass Geld heute investiert werden kann und dadurch Zinsen erwirtschaftet.

In diesem Notebook betrachten wir zwei zentrale Konzepte:

- Present Value (PV) – Barwert
- Future Value (FV) – Endwert

Anschließend setzen wir die Formeln mit Python um und visualisieren die Ergebnisse.

## 1. Future Value

Der Future Value (FV) beschreibt, wie viel ein heute vorhandener Geldbetrag zu einem zukünftigen Zeitpunkt wert ist. Bei jährlicher Verzinsung gilt:

$$
FV = PV \cdot (1+r)^n
$$

Dabei gilt:

- $PV$ = Present Value bzw. heutiger Wert
- $FV$ = Future Value bzw. zukünftiger Wert
- $r$ = Zinssatz pro Periode
- $n$ = Anzahl der Perioden

Der Faktor

$$
(1+r)^n
$$

beschreibt den Zinseszinseffekt.

```python
PV = 1000
r = 0.05
n = 10

FV = PV * (1 + r)**n

print(f"Future Value: {FV:.2f} €")
```

### Interpretation

Wir investieren heute 1000 € zu einem Zinssatz von 5 % pro Jahr. Nach 10 Jahren beträgt der zukünftige Wert:

$$
FV = 1000 \cdot (1.05)^{10}
$$

Das Ergebnis liegt bei ungefähr 1628,89 €. Der ursprüngliche Betrag von 1.000 € ist also durch die Verzinsung auf etwa 1.629 € angewachsen.

## 2. Present Value

Der Present Value (PV) beantwortet die umgekehrte Frage: Wie viel ist ein zukünftiger Geldbetrag heute wert?

Dazu wird der zukünftige Betrag abgezinst. Die Formel lautet:

$$
PV = \frac{FV}{(1+r)^n}
$$

Je weiter ein Geldbetrag in der Zukunft liegt, desto stärker wird er bei einem positiven Zinssatz abgezinst.

```python
FV = 1628.89
r = 0.05
n = 10

PV = FV / (1 + r)**n

print(f"Present Value: {PV:.2f} €")
```

### Interpretation

Ein Betrag von ungefähr 1628,89 € in 10 Jahren hat bei einem Zinssatz von 5 % heute einen Wert von ungefähr 1000 €. Damit sehen wir, dass Present Value und Future Value direkt miteinander verbunden sind:

$$
FV = PV(1+r)^n
$$

und

$$
PV = \frac{FV}{(1+r)^n}
$$

## 3. Der Diskontfaktor

Die Abzinsung kann auch über den sogenannten Diskontfaktor dargestellt werden. Der Diskontfaktor pro Periode ist definiert als:

$$
v = \frac{1}{1+r}
$$

Damit lässt sich der Present Value schreiben als:

$$
PV = FV \cdot v^n
$$

```python
r = 0.05

v = 1 / (1 + r)

print(f"Diskontfaktor: {v:.6f}")
```

## 4. Entwicklung des Future Value über die Zeit

Der Zinseszinseffekt wird besonders deutlich, wenn wir betrachten, wie sich eine Investition über mehrere Jahre entwickelt. Wir investieren:

$$
PV = 1.000
$$

bei einem Zinssatz von:

$$
r = 0.05
$$

und betrachten einen Zeitraum von 30 Jahren.

```python
import numpy as np
import matplotlib.pyplot as plt

PV = 1000
r = 0.05

years = np.arange(0, 31)

future_values = PV * (1 + r)**years

plt.figure(figsize=(10, 5))

plt.plot(years, future_values)

plt.xlabel("Jahre")
plt.ylabel("Future Value (€)")
plt.title("Future Value bei 5 % Verzinsung")

plt.grid(True)
plt.show()
```

Die Kurve zeigt, dass der Future Value nicht linear, sondern exponentiell wächst. Der Grund dafür ist der Zinseszinseffekt: Zinsen werden selbst wieder verzinst.
Je länger das Geld investiert bleibt, desto stärker wirkt dieser Effekt.
