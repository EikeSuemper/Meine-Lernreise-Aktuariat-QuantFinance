# Lists und Arrays in Python

In diesem Notebook lernen wir die wichtigsten Grundlagen zum Arbeiten mit Listen und Arrays in Python.

## 1. Python Lists

Eine Liste ist eine der wichtigsten Datenstrukturen in Python. Mit einer Liste können mehrere Werte unter einem gemeinsamen Namen gespeichert werden.

Eine Liste wird mit eckigen Klammern `[]` erstellt:

```python
meine_liste = [10, 20, 30, 40, 50]
```

Eine Liste kann verschiedene Datentypen enthalten. Zum Beispiel können Zahlen, Texte und Wahrheitswerte gemeinsam gespeichert werden. Listen sind außerdem veränderbar. Das bedeutet, dass einzelne Elemente nachträglich geändert werden können.


````python
# Eine einfache Liste
zahlen = [10, 20, 30, 40, 50]

print(zahlen)
print(type(zahlen))
````

## 2. Data Types

Jeder Wert in Python besitzt einen Datentyp. Einige wichtige Datentypen sind:

| Datentyp | Bedeutung     | Beispiel    |
| -------- | ------------- | ----------- |
| `int`    | Ganze Zahl    | `10`        |
| `float`  | Kommazahl     | `10.5`      |
| `str`    | Text          | `"Hallo"`   |
| `bool`   | Wahrheitswert | `True`      |
| `list`   | Liste         | `[1, 2, 3]` |

Der Datentyp eines Objekts kann mit `type()` überprüft werden. Eine Pythonliste kann Elemente unterschiedlicher Datentypen enthalten.


````python
# Verschiedene Datentypen

alter = 25
gehalt = 3500.50
name = "Max"
ist_versichert = True

print(type(alter))
print(type(gehalt))
print(type(name))
print(type(ist_versichert))
````

````python
# Eine Liste mit unterschiedlichen Datentypen

daten = [25, 3500.50, "Max", True]

print(daten)

for element in daten:
    print(element, type(element))
````

## 3. Indexing

Die einzelnen Elemente einer Liste können über ihren Index angesprochen werden. Wichtig: Python beginnt bei 0 zu zählen.

Bei

```python
zahlen = [10, 20, 30, 40, 50]
```

gilt:

| I ndex | Wert |
| -----: | ---: |
|    `0` | `10` |
|    `1` | `20` |
|    `2` | `30` |
|    `3` | `40` |
|    `4` | `50` |

Das erste Element hat also immer den Index 0. Mit negativen Indizes kann man vom Ende der Liste aus zugreifen:

* `-1` → letztes Element
* `-2` → vorletztes Element
* `-3` → drittletztes Element


````python
zahlen = [10, 20, 30, 40, 50]

print(zahlen[0])
print(zahlen[2])
print(zahlen[4])
````

````python
# Negative Indizes

print(zahlen[-1])  # letztes Element
print(zahlen[-2])  # vorletztes Element
print(zahlen[-3])  # drittletztes Element
````

## 4. Elemente verändern

Listen sind*mutable. Das bedeutet, dass wir bestehende Elemente verändern können. Dazu verwenden wir den Index:

```python
liste[index] = neuer_wert
```

Beispiel:

```python
zahlen[0] = 100
```

Das erste Element wird dadurch von `10` auf `100` geändert.


````python
zahlen = [10, 20, 30, 40, 50]

zahlen[0] = 100

print(zahlen)
````

## 5. Slicing

Mit Slicing können wir mehrere Elemente gleichzeitig aus einer Liste auswählen. Die allgemeine Syntax lautet:

```python
liste[start:stop]
```

Dabei gilt:

* `start` → Startindex, **inklusive**
* `stop` → Endindex, **exklusive**

Beispiel:

```python
zahlen[1:4]
```

liefert die Elemente mit den Indizes `1`, `2` und `3`. Das Element am Index `4` ist nicht enthalten.

### Weitere Möglichkeiten

```python
zahlen[:3]     # erste drei Elemente
zahlen[2:]     # ab Index 2 bis zum Ende
zahlen[:]      # gesamte Liste
zahlen[::2]    # jedes zweite Element
zahlen[::-1]   # Liste umkehren
```


````python
zahlen = [10, 20, 30, 40, 50, 60, 70]

print(zahlen[1:4])
print(zahlen[:3])
print(zahlen[2:])
print(zahlen[:])
````

````python
# Schrittweite

print(zahlen[::2])

# Liste umkehren

print(zahlen[::-1])
````

## 6. Daten zu einer Liste hinzufügen

Es gibt verschiedene Möglichkeiten, Daten zu einer Liste hinzuzufügen.

### `append()`

`append()` fügt ein Element am Ende der Liste hinzu.

```python
zahlen.append(80)
```

### `insert()`

Mit `insert()` kann ein Element an einer bestimmten Position eingefügt werden.

```python
zahlen.insert(2, 25)
```

Dabei wird `25` an Index `2` eingefügt.

### `extend()`

Mit `extend()` können mehrere Elemente hinzugefügt werden.

```python
zahlen.extend([80, 90, 100])
```


````python
zahlen = [10, 20, 30]

# Ein Element hinzufügen
zahlen.append(40)

print(zahlen)
````

````python
zahlen = [10, 20, 30]

# Element an bestimmter Position einfügen
zahlen.insert(1, 15)

print(zahlen)
````

````python
zahlen = [10, 20, 30]

# Mehrere Elemente hinzufügen
zahlen.extend([40, 50, 60])

print(zahlen)
````
